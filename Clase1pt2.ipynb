Diccionarios y probabilidad

Lanzamiento de 3 monedas


```python
from itertools import product
Omega = set(product(["A", "S"], repeat=3))
Omega
```

Consideremos el evento E: la cara de las tres monedas es "aguila". En este caso, si detona por A al evento: el resultado del i-enesimo lanzamiento es aguila, para i= 1,2,3, entonces

Representamos los conjuntos A:

```python
A_1 = {om for om in Omega if om[0] == "A"}
A_1
```

```python
A_2 = {om for om in Omega if om[1] == "A"}
A_2
```

```python
A_3 = {om for om in Omega if om[2] == "A"}
A_3
```

El evento E tambien puede ser representado usando el metodo .count

```python
E = {om for om in Omega if om.count("A") == 3}
E
```

```python
from fractions import Fraction

def P(A):
    P = Fraction(len(A), len(Omega))
    return P
```

Por lo tanto,

```python
P(E)
```

En  esta situacion, podemos calcular la probabilidad del evento F: cae aguila en el primer y sehundo lanzamiento.

```python
F = A_1.intersection(A_2.intersection(Omega))
P(F)
```

Lanzamiento de monedas injustas (Voladas)

```python
import random
```

```python
def volado(p):
    return 'sol' if random.random() < p else 'aguila'
```

Por ejemplo, si la moneda tiene carga en A igual a 0.4, podemos realizar una sucesion de lanzamientoy obtener las veces que cae 'Aguila'

```python
volado(0.4)
```

```python
a= [volado(0.4) for i in range(10)]
a
```

```python
resultados =[]

for i in range(10):
    resultado = volado(0.7)
    resultados.append(resultado)
resultados
```

```python
resultados.count('sol')
```

```python
p = []

for i in range(1000):

    N = 10
    resultados = []

    for i in range(N):
        resultado = volado(0.8)
        resultados.append(resultado) 

    n_soles = resultados.count('sol')  
    p_soles = n_soles / N
    p.append(p_soles)

#probabilidad promedio
p_soles_ = sum(p) / 1000

print('Probabilidad {:.3f}'.format(p_soles_))
```
